```python
def make_list(predicted_periods, regression_results):
    results = []
    for i in range(len(predicted_periods)):
        count = [ -1 for i in range(np.argmax(regression_results[i][1][0]))]
        results.append([predicted_periods[i], count])
    return results
```

```python
A_list = make_list(predicted_periods, regression_results)
print(A_list)
```

```python
for i in range(len(regression_AB_results)):
    print(len(regression_AB_results[i]))
    print(f"regression AB result: {regression_AB_results[i]}")
```

```python
total_AB_results = []
A_result = 0
B_result = 0
for result in regression_AB_results:
    num = len(result)//2
    A_param = 0
    B_param = 0
    for i in range(num):
        A_param += result[i*2]
        B_param += result[i*2+1]
    print(f"A_param: {A_param/num}, B_param: {B_param/num}")
    total_AB_results.append([A_param/num, B_param/num])
total_AB_results = np.array(total_AB_results)
print(f"[total_AB_results] A:{np.mean(total_AB_results[:,0])}, B:{np.mean(total_AB_results[:,1])}")
```
